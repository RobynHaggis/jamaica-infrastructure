```python
landcover10kgrid["area_ha"] = landcover10kgrid.geometry.area * 0.0001  # m2 to ha
```

```python
landcover_area_by_cell = landcover10kgrid[["Classify", "index_i", "index_j", "area_ha"]] \
    .groupby(["index_i", "index_j", "Classify"]) \
    .sum()

landcover_area_by_cell
```

```python
area_by_cell = landcover10kgrid[["index_i", "index_j", "area_ha"]] \
    .groupby(["index_i", "index_j"]) \
    .sum() \
    .rename(columns={"area_ha": "cell_area_ha"})

area_by_cell
```

```python
class_area = landcover_area_by_cell.join(area_by_cell)
class_area
```

```python
class_area["area_fraction"] =  class_area.area_ha / class_area.cell_area_ha
```

```python
class_area
```

```python
landcover_with_fraction = landcover10kgrid.set_index(["index_i", "index_j", "Classify"]).join(class_area[["area_fraction"]]).reset_index()
```

```python
landcover_with_fraction.plot(column="area_fraction")
```

```python
def count_cells(df):
    return df[["index_i","index_j"]].drop_duplicates().size
```

```python
extract = landcover_with_fraction[
    (landcover_with_fraction["Classify"]=="Closed broadleaved forest (Primary Forest)")
    & (landcover_with_fraction["area_fraction"] > 0.5)
]
#extract.plot(column="area_fraction")

count_cells(extract), count_cells(landcover_with_fraction)
```
