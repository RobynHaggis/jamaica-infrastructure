# Landslide step 06: expected annual damage maps (minimum_scenario)

- Loads asset-level EAD output from step 05.
- Maps avoided EAD from restoration and protection scenarios.
- Saves map layers and PNG maps for reporting.

```python
from pathlib import Path

import numpy as np
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
from matplotlib.colors import TwoSlopeNorm, LinearSegmentedColormap
from matplotlib.cm import ScalarMappable

pd.set_option('display.max_columns', 200)
pd.set_option('display.width', 200)

try:
    from IPython.display import display
except Exception:
    def display(x):
        print(x)
```

```python
# Core paths
base_path = Path('/Users/robynhaggis/Documents/Geospatial_analysis/dphil_papers')
results_path = base_path / 'dphil_paper_3/results/02_damage_estimates/landslide_damages/results_landslide_minimum_scenario'
output_damage_estimates = results_path / 'damage_estimates'

asset_ead_file = output_damage_estimates / 'landslide_ead_asset_level_usd.csv'
network_csv = base_path / 'dphil_common_cross_cutting/common_incoming_data/networks/network_layers_hazard_intersections_details.csv'
jamaica_boundary_path = base_path / 'dphil_common_cross_cutting/common_incoming_data/boundaries/jamaica.gpkg'

data_root = base_path / 'dphil_common_cross_cutting/common_incoming_data'

for required in [asset_ead_file, network_csv, jamaica_boundary_path]:
    if not required.exists():
        raise FileNotFoundError(f'Missing required file: {required}')

print('Asset EAD file:', asset_ead_file)
print('Network csv:', network_csv)
print('Jamaica boundary:', jamaica_boundary_path)
```

```python
# Load EAD output and compute avoided metrics of interest
asset_ead = pd.read_csv(asset_ead_file)
required_cols = [
    'Sector', 'Subsector', 'Asset', 'Layer', 'Asset_ID',
    'EAD_Baseline_USD', 'EAD_Deforestation_USD', 'EAD_Reforestation_USD'
]
missing_cols = [c for c in required_cols if c not in asset_ead.columns]
if missing_cols:
    raise KeyError(f'Missing required columns in asset EAD file: {missing_cols}')

asset_ead = asset_ead.copy()
asset_ead['Avoided_EAD_Restoration_USD'] = asset_ead['EAD_Baseline_USD'] - asset_ead['EAD_Reforestation_USD']
asset_ead['Avoided_EAD_Protection_USD'] = asset_ead['EAD_Deforestation_USD'] - asset_ead['EAD_Baseline_USD']
asset_ead['Combined_Benefit_Reforestation_vs_Deforestation_USD'] = asset_ead['EAD_Deforestation_USD'] - asset_ead['EAD_Reforestation_USD']

print(f'Loaded asset EAD rows: {len(asset_ead):,}')
display(asset_ead.head(10))
```

```python
# Build geospatial asset map layer by joining asset geometries with asset-level EAD metrics
network_details = pd.read_csv(network_csv)
network_details = network_details[[
    'sector', 'asset_description', 'asset_gpkg', 'asset_layer', 'asset_id_column', 'path'
]].drop_duplicates().copy()


def resolve_network_asset_file(asset_relative_path):
    relative_asset_path = Path(asset_relative_path)
    asset_file_in_common_incoming_data = data_root / relative_asset_path
    asset_file_in_nested_networks_folder = data_root / 'networks' / relative_asset_path

    if asset_file_in_common_incoming_data.exists():
        return asset_file_in_common_incoming_data
    if asset_file_in_nested_networks_folder.exists():
        return asset_file_in_nested_networks_folder

    raise FileNotFoundError(
        f"Could not find asset file '{relative_asset_path}'. Checked: {asset_file_in_common_incoming_data} ; {asset_file_in_nested_networks_folder}"
    )


include_zero_assets = False
map_layers = []
missing_asset_files = []

for row in network_details.itertuples(index=False):
    ead_subset = asset_ead.loc[
        (asset_ead['Asset'] == row.asset_gpkg) & (asset_ead['Layer'] == row.asset_layer),
        [
            'Asset_ID',
            'EAD_Baseline_USD', 'EAD_Deforestation_USD', 'EAD_Reforestation_USD',
            'Avoided_EAD_Restoration_USD', 'Avoided_EAD_Protection_USD',
            'Combined_Benefit_Reforestation_vs_Deforestation_USD'
        ]
    ].copy()

    if ead_subset.empty:
        continue

    if not include_zero_assets:
        ead_subset = ead_subset[
            (ead_subset['Avoided_EAD_Restoration_USD'] != 0)
            | (ead_subset['Avoided_EAD_Protection_USD'] != 0)
        ].copy()

    if ead_subset.empty:
        continue

    try:
        asset_file = resolve_network_asset_file(row.path)
    except FileNotFoundError:
        missing_asset_files.append(row.path)
        continue

    asset_gdf = gpd.read_file(asset_file, layer=row.asset_layer)
    if row.asset_id_column not in asset_gdf.columns:
        print(f"Skipping {row.asset_gpkg}_{row.asset_layer}: missing id column '{row.asset_id_column}'")
        continue

    asset_gdf = asset_gdf[[row.asset_id_column, 'geometry']].copy()
    asset_gdf = gpd.GeoDataFrame(asset_gdf, geometry='geometry', crs=asset_gdf.crs)
    if asset_gdf.crs is not None:
        asset_gdf = asset_gdf.to_crs('EPSG:3448')

    asset_gdf['_join_id'] = asset_gdf[row.asset_id_column].astype(str)
    ead_subset['_join_id'] = ead_subset['Asset_ID'].astype(str)

    merged = asset_gdf.merge(ead_subset, on='_join_id', how='inner')
    if merged.empty:
        continue

    merged['Sector'] = row.sector
    merged['Subsector'] = row.asset_description
    merged['Asset'] = row.asset_gpkg
    merged['Layer'] = row.asset_layer

    map_layers.append(merged[[
        'Sector', 'Subsector', 'Asset', 'Layer', row.asset_id_column,
        'Asset_ID',
        'EAD_Baseline_USD', 'EAD_Deforestation_USD', 'EAD_Reforestation_USD',
        'Avoided_EAD_Restoration_USD', 'Avoided_EAD_Protection_USD',
        'Combined_Benefit_Reforestation_vs_Deforestation_USD',
        'geometry'
    ]].rename(columns={row.asset_id_column: 'Asset_ID_Source'}))

if not map_layers:
    raise ValueError('No map features were built. Check step-05 outputs and asset files.')

landslide_ead_map_layers = gpd.GeoDataFrame(
    pd.concat(map_layers, ignore_index=True),
    geometry='geometry',
    crs='EPSG:3448'
)

map_layers_file = output_damage_estimates / 'landslide_ead_asset_map_layers.gpkg'
landslide_ead_map_layers.to_file(map_layers_file, driver='GPKG')

print(f'Map features loaded: {len(landslide_ead_map_layers):,}')
print('Features by sector:')
display(landslide_ead_map_layers.groupby('Sector', as_index=False).size())
print('Saved map layer:', map_layers_file)

if missing_asset_files:
    print('Missing asset files skipped:')
    for p in sorted(set(missing_asset_files)):
        print('-', p)
```

```python
# Overall and sector totals for avoided EAD metrics (USD + readable labels)

def format_usd_readable(value):
    if pd.isna(value):
        return 'NA'
    abs_value = abs(float(value))
    sign = '-' if float(value) < 0 else ''
    if abs_value >= 1_000_000_000:
        return f"{sign}${abs_value / 1_000_000_000:,.2f} billion"
    if abs_value >= 1_000_000:
        return f"{sign}${abs_value / 1_000_000:,.2f} million"
    if abs_value >= 1_000:
        return f"{sign}${abs_value / 1_000:,.2f} thousand"
    return f"{sign}${abs_value:,.2f}"

summary_cols = [
    'Avoided_EAD_Restoration_USD',
    'Avoided_EAD_Protection_USD',
    'Combined_Benefit_Reforestation_vs_Deforestation_USD'
]

overall_summary = pd.DataFrame([landslide_ead_map_layers[summary_cols].sum()])
overall_summary['Avoided_EAD_Restoration_Readable'] = overall_summary['Avoided_EAD_Restoration_USD'].apply(format_usd_readable)
overall_summary['Avoided_EAD_Protection_Readable'] = overall_summary['Avoided_EAD_Protection_USD'].apply(format_usd_readable)
overall_summary['Combined_Benefit_Readable'] = overall_summary['Combined_Benefit_Reforestation_vs_Deforestation_USD'].apply(format_usd_readable)

sector_summary = landslide_ead_map_layers.groupby('Sector', as_index=False)[summary_cols].sum()
sector_summary['Avoided_EAD_Restoration_Readable'] = sector_summary['Avoided_EAD_Restoration_USD'].apply(format_usd_readable)
sector_summary['Avoided_EAD_Protection_Readable'] = sector_summary['Avoided_EAD_Protection_USD'].apply(format_usd_readable)
sector_summary['Combined_Benefit_Readable'] = sector_summary['Combined_Benefit_Reforestation_vs_Deforestation_USD'].apply(format_usd_readable)

overall_file = output_damage_estimates / 'landslide_ead_avoided_overall_totals_usd.csv'
sector_file = output_damage_estimates / 'landslide_ead_avoided_sector_totals_usd.csv'
overall_summary.to_csv(overall_file, index=False)
sector_summary.to_csv(sector_file, index=False)

print('Overall totals:')
display(overall_summary)
print('Sector totals:')
display(sector_summary.sort_values('Sector'))
print('Saved:', overall_file)
print('Saved:', sector_file)
```

```python
# Subsector breakdown table (within-sector drivers)
subsector_summary = (
    landslide_ead_map_layers
    .groupby(['Sector', 'Subsector'], as_index=False)[summary_cols]
    .sum()
)

# Shares of sector totals by metric to show what drives each sector result
sector_totals_for_share = sector_summary.set_index('Sector')[summary_cols]

for metric_col, share_col in [
    ('Avoided_EAD_Restoration_USD', 'Restoration_Share_of_Sector'),
    ('Avoided_EAD_Protection_USD', 'Protection_Share_of_Sector'),
    ('Combined_Benefit_Reforestation_vs_Deforestation_USD', 'Combined_Benefit_Share_of_Sector'),
]:
    subsector_summary[share_col] = np.nan
    for sector_name in subsector_summary['Sector'].unique():
        denom = float(sector_totals_for_share.loc[sector_name, metric_col]) if sector_name in sector_totals_for_share.index else 0.0
        mask = subsector_summary['Sector'] == sector_name
        if denom != 0:
            subsector_summary.loc[mask, share_col] = subsector_summary.loc[mask, metric_col] / denom
        else:
            subsector_summary.loc[mask, share_col] = pd.NA

subsector_summary['Avoided_EAD_Restoration_Readable'] = subsector_summary['Avoided_EAD_Restoration_USD'].apply(format_usd_readable)
subsector_summary['Avoided_EAD_Protection_Readable'] = subsector_summary['Avoided_EAD_Protection_USD'].apply(format_usd_readable)
subsector_summary['Combined_Benefit_Readable'] = subsector_summary['Combined_Benefit_Reforestation_vs_Deforestation_USD'].apply(format_usd_readable)

subsector_summary['Restoration_Share_of_Sector_Pct'] = (100.0 * subsector_summary['Restoration_Share_of_Sector']).round(2)
subsector_summary['Protection_Share_of_Sector_Pct'] = (100.0 * subsector_summary['Protection_Share_of_Sector']).round(2)
subsector_summary['Combined_Benefit_Share_of_Sector_Pct'] = (100.0 * subsector_summary['Combined_Benefit_Share_of_Sector']).round(2)

subsector_file = output_damage_estimates / 'landslide_ead_avoided_subsector_totals_usd.csv'
subsector_summary.to_csv(subsector_file, index=False)

print('Subsector totals (drivers within each sector):')
display(subsector_summary.sort_values(['Sector', 'Subsector']))
print('Saved:', subsector_file)
```

```python
# Grouped bar chart: sector and total EAD by scenario (USD million)
chart_sector_order = ['buildings', 'transport', 'water', 'energy']
scenario_cols = [
    ('EAD_Baseline_USD', 'Baseline'),
    ('EAD_Deforestation_USD', 'Deforestation'),
    ('EAD_Reforestation_USD', 'Reforestation'),
]

sector_chart = (
    asset_ead
    .groupby('Sector', as_index=False)[[c for c, _ in scenario_cols]]
    .sum()
)

# Ensure all expected sectors are present and in a stable order.
sector_chart = sector_chart.set_index('Sector')
for sector_name in chart_sector_order:
    if sector_name not in sector_chart.index:
        sector_chart.loc[sector_name, [c for c, _ in scenario_cols]] = 0.0
sector_chart = sector_chart.loc[chart_sector_order]

total_row = pd.DataFrame(
    {
        c: [float(sector_chart[c].sum())]
        for c, _ in scenario_cols
    },
    index=['TOTAL'],
)

sector_chart = pd.concat([sector_chart, total_row], axis=0)
sector_chart_musd = sector_chart / 1_000_000.0

x = np.arange(len(sector_chart_musd.index))
bar_width = 0.24

scenario_styles = {
    'Baseline': {'color': '#b3b3b3', 'hatch': ''},
    'Deforestation': {'color': '#f4a259', 'hatch': '//'},
    'Reforestation': {'color': '#6dbf6b', 'hatch': '///'},
}

fig, ax = plt.subplots(figsize=(11, 6))

for idx, (col, label) in enumerate(scenario_cols):
    offsets = x + (idx - 1) * bar_width
    ax.bar(
        offsets,
        sector_chart_musd[col].to_numpy(),
        width=bar_width,
        label=label,
        color=scenario_styles[label]['color'],
        edgecolor='#3a3a3a',
        linewidth=0.8,
        hatch=scenario_styles[label]['hatch'],
    )

if len(sector_chart_musd.index) >= 2:
    ax.axvline(len(sector_chart_musd.index) - 1.5, linestyle=':', color='#808080', linewidth=1.1)

xticklabels = [s.capitalize() if s != 'TOTAL' else 'TOTAL' for s in sector_chart_musd.index]
ax.set_xticks(x)
ax.set_xticklabels(xticklabels)
ax.set_ylabel('EAD (USD million)')
ax.set_title('Landslide EAD by Sector and Scenario')
ax.grid(axis='y', alpha=0.25)
ax.legend(frameon=True, ncol=3, loc='upper left')

plt.tight_layout()

scenario_chart_file = output_damage_estimates / 'landslide_sector_total_ead_usd_by_scenario_grouped.png'
fig.savefig(scenario_chart_file, dpi=300, bbox_inches='tight')
print('Saved chart:', scenario_chart_file)
plt.show()

display(sector_chart.reset_index().rename(columns={'index': 'Sector'}))
```

```python
# Stacked bar chart: baseline with reforestation and deforestation deltas (USD million)
stack_sector_order = ['buildings', 'transport', 'water', 'energy']

stack_df = (
    asset_ead
    .groupby('Sector', as_index=False)[['EAD_Baseline_USD', 'EAD_Reforestation_USD', 'EAD_Deforestation_USD']]
    .sum()
    .set_index('Sector')
)

for sector_name in stack_sector_order:
    if sector_name not in stack_df.index:
        stack_df.loc[sector_name, ['EAD_Baseline_USD', 'EAD_Reforestation_USD', 'EAD_Deforestation_USD']] = 0.0

stack_df = stack_df.loc[stack_sector_order]
stack_df.loc['TOTAL', 'EAD_Baseline_USD'] = float(stack_df['EAD_Baseline_USD'].sum())
stack_df.loc['TOTAL', 'EAD_Reforestation_USD'] = float(stack_df['EAD_Reforestation_USD'].sum())
stack_df.loc['TOTAL', 'EAD_Deforestation_USD'] = float(stack_df['EAD_Deforestation_USD'].sum())

stack_df['Delta_Reforestation_vs_Baseline_USD'] = stack_df['EAD_Baseline_USD'] - stack_df['EAD_Reforestation_USD']
stack_df['Delta_Deforestation_vs_Baseline_USD'] = stack_df['EAD_Deforestation_USD'] - stack_df['EAD_Baseline_USD']

stack_df['Baseline_USD_million'] = stack_df['EAD_Baseline_USD'] / 1_000_000.0
stack_df['Reforestation_Delta_USD_million'] = stack_df['Delta_Reforestation_vs_Baseline_USD'] / 1_000_000.0
stack_df['Deforestation_Delta_USD_million'] = stack_df['Delta_Deforestation_vs_Baseline_USD'] / 1_000_000.0

stack_df['Reforestation_Delta_Share_Pct'] = np.where(
    stack_df['EAD_Baseline_USD'] > 0,
    100.0 * stack_df['Delta_Reforestation_vs_Baseline_USD'] / stack_df['EAD_Baseline_USD'],
    np.nan,
)
stack_df['Deforestation_Delta_Share_Pct'] = np.where(
    stack_df['EAD_Baseline_USD'] > 0,
    100.0 * stack_df['Delta_Deforestation_vs_Baseline_USD'] / stack_df['EAD_Baseline_USD'],
    np.nan,
)

x = np.arange(len(stack_df.index))
baseline_vals = stack_df['Baseline_USD_million'].to_numpy()
ref_delta = stack_df['Reforestation_Delta_USD_million'].to_numpy()
def_delta = stack_df['Deforestation_Delta_USD_million'].to_numpy()

ref_pos = np.clip(ref_delta, 0, None)
ref_neg = np.clip(ref_delta, None, 0)
def_pos = np.clip(def_delta, 0, None)
def_neg = np.clip(def_delta, None, 0)

fig, ax = plt.subplots(figsize=(11.8, 6.8))

ax.bar(
    x,
    baseline_vals,
    color='#c0c0c0',
    edgecolor='#4d4d4d',
    linewidth=0.8,
    label='Baseline EAD',
)

# Stack positive deltas above baseline (green then orange).
ax.bar(
    x,
    ref_pos,
    bottom=baseline_vals,
    color='#d9f0d3',
    edgecolor='#2b8c3e',
    linewidth=0.8,
    hatch='///',
    label='Delta vs baseline: Reforestation (Baseline - Reforestation)',
)
ax.bar(
    x,
    def_pos,
    bottom=baseline_vals + ref_pos,
    color='#fdd49e',
    edgecolor='#d95f0e',
    linewidth=0.8,
    hatch='xx',
    label='Delta vs baseline: Deforestation (Deforestation - Baseline)',
)

# Stack negative deltas below baseline if present.
if (ref_neg < 0).any():
    ax.bar(
        x,
        ref_neg,
        bottom=baseline_vals,
        color='#c7e9c0',
        edgecolor='#238b45',
        linewidth=0.8,
        hatch='\\',
        label='Negative reforestation delta',
    )
if (def_neg < 0).any():
    ax.bar(
        x,
        def_neg,
        bottom=baseline_vals + ref_neg,
        color='#deebf7',
        edgecolor='#3182bd',
        linewidth=0.8,
        hatch='..',
        label='Negative deforestation delta',
    )

if len(stack_df.index) >= 2:
    ax.axvline(len(stack_df.index) - 1.5, linestyle=':', color='#808080', linewidth=1.1)

xticklabels = [s.capitalize() if s != 'TOTAL' else 'TOTAL' for s in stack_df.index]
ax.set_xticks(x)
ax.set_xticklabels(xticklabels)
ax.set_ylabel('EAD (USD million)')
ax.set_title('Baseline EAD with Reforestation and Deforestation Deltas')
ax.grid(axis='y', alpha=0.25)

handles, labels = ax.get_legend_handles_labels()
seen = set()
uniq_h, uniq_l = [], []
for h, l in zip(handles, labels):
    if l not in seen and l:
        seen.add(l)
        uniq_h.append(h)
        uniq_l.append(l)
ax.legend(uniq_h, uniq_l, frameon=True, loc='upper left')

for xi, base, refp, defp in zip(x, baseline_vals, stack_df['Reforestation_Delta_Share_Pct'].to_numpy(), stack_df['Deforestation_Delta_Share_Pct'].to_numpy()):
    if np.isfinite(refp):
        ax.text(xi, base, f"R {refp:+.1f}%", ha='center', va='bottom', fontsize=8, color='#238b45')
    if np.isfinite(defp):
        ax.text(xi, base * 0.985, f"D {defp:+.1f}%", ha='center', va='top', fontsize=8, color='#d95f0e')

plt.tight_layout()

stacked_chart_file = output_damage_estimates / 'landslide_baseline_with_scenario_deltas_stacked.png'
fig.savefig(stacked_chart_file, dpi=300, bbox_inches='tight')
print('Saved chart:', stacked_chart_file)
plt.show()

display(
    stack_df.reset_index().rename(columns={'index': 'Sector'})[
        [
            'Sector',
            'EAD_Baseline_USD', 'EAD_Reforestation_USD', 'EAD_Deforestation_USD',
            'Delta_Reforestation_vs_Baseline_USD', 'Delta_Deforestation_vs_Baseline_USD',
            'Reforestation_Delta_Share_Pct', 'Deforestation_Delta_Share_Pct'
        ]
    ]
)
```

```python
# Mapping helpers
jamaica_boundary = gpd.read_file(jamaica_boundary_path).to_crs('EPSG:3448')
map_out_dir = output_damage_estimates / 'maps_landslide_avoided_ead'
map_out_dir.mkdir(parents=True, exist_ok=True)

cmap = LinearSegmentedColormap.from_list(
    'red_white_green',
    ['#c81e1e', '#ffffff', '#0b8f3f'],
    N=256,
)

def choose_unit(abs_max_usd):
    if abs_max_usd >= 1e6:
        return 1e6, 'USD millions'
    if abs_max_usd >= 1e3:
        return 1e3, 'USD thousands'
    return 1.0, 'USD'


def plot_metric_map(gdf, value_col, title, output_png, display_quantile=0.995):
    values_usd = pd.to_numeric(gdf[value_col], errors='coerce').fillna(0.0)
    abs_values = values_usd.abs()
    cap_usd = float(abs_values.quantile(display_quantile))
    if cap_usd <= 0:
        cap_usd = float(abs_values.max()) if float(abs_values.max()) > 0 else 1.0

    unit_factor, unit_label = choose_unit(cap_usd)
    cap = cap_usd / unit_factor

    plot_gdf = gdf.copy()
    plot_gdf['_plot_val'] = (values_usd / unit_factor).clip(-cap, cap)

    norm = TwoSlopeNorm(vmin=-cap, vcenter=0.0, vmax=cap)

    fig, ax = plt.subplots(figsize=(11, 10))
    ax.set_facecolor('#ffffff')
    jamaica_boundary.boundary.plot(ax=ax, color='#bdbdbd', linewidth=0.45, zorder=1)

    geom_type = plot_gdf.geometry.geom_type.astype(str)
    polys = plot_gdf[geom_type.str.contains('Polygon', na=False)]
    lines = plot_gdf[geom_type.str.contains('LineString', na=False)]
    points = plot_gdf[geom_type.str.contains('Point', na=False)]

    if not polys.empty:
        polys.plot(ax=ax, column='_plot_val', cmap=cmap, norm=norm, linewidth=0.12, edgecolor='none', alpha=0.9, zorder=2)
    if not lines.empty:
        lines.plot(ax=ax, column='_plot_val', cmap=cmap, norm=norm, linewidth=0.9, alpha=0.95, zorder=3)
    if not points.empty:
        points.plot(ax=ax, column='_plot_val', cmap=cmap, norm=norm, markersize=14, alpha=0.95, zorder=4)

    sm = ScalarMappable(norm=norm, cmap=cmap)
    sm.set_array([])
    cbar = fig.colorbar(sm, ax=ax, fraction=0.036, pad=0.02)
    cbar.set_label(f'{value_col} ({unit_label}), clipped at q={display_quantile:.3f}', rotation=90)

    ax.set_title(title, fontsize=13)
    ax.set_axis_off()
    plt.tight_layout()

    fig.savefig(output_png, dpi=300, bbox_inches='tight')
    print('Saved:', output_png)
    plt.show()
```

```python
# Forest-associated increase maps (where forested scenarios have higher damages)
from matplotlib.colors import Normalize

# Positive values represent locations where damages are higher with forests.
# 1) Protection comparison (Deforestation - Baseline):
#    if negative, then Baseline > Deforestation => forests associated with higher damages.
# 2) Restoration comparison (Baseline - Reforestation):
#    if negative, then Reforestation > Baseline => restoration associated with higher damages.

landslide_ead_map_layers['Forest_Associated_Increase_Protection_USD'] = (
    -pd.to_numeric(landslide_ead_map_layers['Avoided_EAD_Protection_USD'], errors='coerce').fillna(0.0)
).clip(lower=0.0)

landslide_ead_map_layers['Forest_Associated_Increase_Restoration_USD'] = (
    -pd.to_numeric(landslide_ead_map_layers['Avoided_EAD_Restoration_USD'], errors='coerce').fillna(0.0)
).clip(lower=0.0)


def plot_positive_only_map(gdf, value_col, title, output_png, display_quantile=0.995):
    values_usd = pd.to_numeric(gdf[value_col], errors='coerce').fillna(0.0)
    positive_mask = values_usd > 0

    if positive_mask.sum() == 0:
        print(f'No features with positive {value_col}; no map generated: {output_png.name}')
        return False

    plot_gdf = gdf.loc[positive_mask].copy()
    values_usd = values_usd.loc[positive_mask]

    cap_usd = float(values_usd.quantile(display_quantile))
    if cap_usd <= 0:
        cap_usd = float(values_usd.max()) if float(values_usd.max()) > 0 else 1.0

    unit_factor, unit_label = choose_unit(cap_usd)
    cap = cap_usd / unit_factor

    plot_gdf['_plot_val'] = (values_usd / unit_factor).clip(0, cap)
    norm = Normalize(vmin=0, vmax=cap)

    increase_cmap = LinearSegmentedColormap.from_list('white_red', ['#ffffff', '#cb181d'], N=256)

    fig, ax = plt.subplots(figsize=(11, 10))
    ax.set_facecolor('#ffffff')
    jamaica_boundary.boundary.plot(ax=ax, color='#bdbdbd', linewidth=0.45, zorder=1)

    geom_type = plot_gdf.geometry.geom_type.astype(str)
    polys = plot_gdf[geom_type.str.contains('Polygon', na=False)]
    lines = plot_gdf[geom_type.str.contains('LineString', na=False)]
    points = plot_gdf[geom_type.str.contains('Point', na=False)]

    if not polys.empty:
        polys.plot(ax=ax, column='_plot_val', cmap=increase_cmap, norm=norm, linewidth=0.12, edgecolor='none', alpha=0.9, zorder=2)
    if not lines.empty:
        lines.plot(ax=ax, column='_plot_val', cmap=increase_cmap, norm=norm, linewidth=0.9, alpha=0.95, zorder=3)
    if not points.empty:
        points.plot(ax=ax, column='_plot_val', cmap=increase_cmap, norm=norm, markersize=14, alpha=0.95, zorder=4)

    sm = ScalarMappable(norm=norm, cmap=increase_cmap)
    sm.set_array([])
    cbar = fig.colorbar(sm, ax=ax, fraction=0.036, pad=0.02)
    cbar.set_label(f'{value_col} ({unit_label}), clipped at q={display_quantile:.3f}', rotation=90)

    ax.set_title(title, fontsize=13)
    ax.set_axis_off()
    plt.tight_layout()

    fig.savefig(output_png, dpi=300, bbox_inches='tight')
    print('Saved:', output_png)
    plt.show()
    return True


# Save asset-level table for colleague review
forest_increase_assets = landslide_ead_map_layers.loc[
    (landslide_ead_map_layers['Forest_Associated_Increase_Protection_USD'] > 0)
    | (landslide_ead_map_layers['Forest_Associated_Increase_Restoration_USD'] > 0),
    [
        'Sector', 'Subsector', 'Asset', 'Layer', 'Asset_ID',
        'EAD_Baseline_USD', 'EAD_Deforestation_USD', 'EAD_Reforestation_USD',
        'Forest_Associated_Increase_Protection_USD',
        'Forest_Associated_Increase_Restoration_USD',
        'geometry'
    ]
].copy()

forest_increase_asset_csv = output_damage_estimates / 'landslide_forest_associated_damage_increase_asset_level_usd.csv'
forest_increase_assets.drop(columns=['geometry']).to_csv(forest_increase_asset_csv, index=False)
print('Saved:', forest_increase_asset_csv)

# Totals and sector totals
forest_increase_totals = pd.DataFrame([{
    'Forest_Associated_Increase_Protection_USD': float(landslide_ead_map_layers['Forest_Associated_Increase_Protection_USD'].sum()),
    'Forest_Associated_Increase_Restoration_USD': float(landslide_ead_map_layers['Forest_Associated_Increase_Restoration_USD'].sum()),
    'Protection_Increase_Asset_Count': int((landslide_ead_map_layers['Forest_Associated_Increase_Protection_USD'] > 0).sum()),
    'Restoration_Increase_Asset_Count': int((landslide_ead_map_layers['Forest_Associated_Increase_Restoration_USD'] > 0).sum()),
}])
forest_increase_totals['Forest_Associated_Increase_Protection_Readable'] = forest_increase_totals['Forest_Associated_Increase_Protection_USD'].apply(format_usd_readable)
forest_increase_totals['Forest_Associated_Increase_Restoration_Readable'] = forest_increase_totals['Forest_Associated_Increase_Restoration_USD'].apply(format_usd_readable)

forest_increase_totals_csv = output_damage_estimates / 'landslide_forest_associated_damage_increase_totals_usd.csv'
forest_increase_totals.to_csv(forest_increase_totals_csv, index=False)
print('Saved:', forest_increase_totals_csv)

forest_increase_sector = (
    landslide_ead_map_layers
    .groupby('Sector', as_index=False)[
        ['Forest_Associated_Increase_Protection_USD', 'Forest_Associated_Increase_Restoration_USD']
    ]
    .sum()
)
forest_increase_sector['Forest_Associated_Increase_Protection_Readable'] = forest_increase_sector['Forest_Associated_Increase_Protection_USD'].apply(format_usd_readable)
forest_increase_sector['Forest_Associated_Increase_Restoration_Readable'] = forest_increase_sector['Forest_Associated_Increase_Restoration_USD'].apply(format_usd_readable)

forest_increase_sector_csv = output_damage_estimates / 'landslide_forest_associated_damage_increase_sector_totals_usd.csv'
forest_increase_sector.to_csv(forest_increase_sector_csv, index=False)
print('Saved:', forest_increase_sector_csv)

# Maps (positive-only)
forest_increase_map_protection = map_out_dir / 'forest_associated_damage_increase_protection_only_all_sectors.png'
forest_increase_map_restoration = map_out_dir / 'forest_associated_damage_increase_restoration_only_all_sectors.png'

has_protection_increase = plot_positive_only_map(
    landslide_ead_map_layers,
    value_col='Forest_Associated_Increase_Protection_USD',
    title='Damage Increase Where Baseline > Deforestation (Forests-associated increase)',
    output_png=forest_increase_map_protection,
)

has_restoration_increase = plot_positive_only_map(
    landslide_ead_map_layers,
    value_col='Forest_Associated_Increase_Restoration_USD',
    title='Damage Increase Where Reforestation > Baseline (Restoration-associated increase)',
    output_png=forest_increase_map_restoration,
)

if not has_restoration_increase:
    no_map_note = output_damage_estimates / 'forest_associated_damage_increase_restoration_none.txt'
    no_map_note.write_text('No assets with positive Forest_Associated_Increase_Restoration_USD in this scenario run.\n', encoding='utf-8')
    print('Saved:', no_map_note)

print('Forest-associated increase totals:')
display(forest_increase_totals)
print('By sector:')
display(forest_increase_sector.sort_values('Sector'))


# High-contrast highlight maps (binary: any increase shown in red)
def plot_binary_increase_map(gdf, value_col, title, output_png):
    values = pd.to_numeric(gdf[value_col], errors='coerce').fillna(0.0)
    inc_mask = values > 0

    if inc_mask.sum() == 0:
        print(f'No positive {value_col}; no highlight map generated: {output_png.name}')
        return False

    inc = gdf.loc[inc_mask].copy()
    total_increase = float(values.loc[inc_mask].sum())

    fig, ax = plt.subplots(figsize=(11, 10))
    ax.set_facecolor('#ffffff')

    # Jamaica outline
    jamaica_boundary.boundary.plot(ax=ax, color='#9e9e9e', linewidth=0.6, zorder=1)

    # Context: all mapped assets in very light grey
    gt_all = gdf.geometry.geom_type.astype(str)
    all_polys = gdf[gt_all.str.contains('Polygon', na=False)]
    all_lines = gdf[gt_all.str.contains('LineString', na=False)]
    all_points = gdf[gt_all.str.contains('Point', na=False)]

    if not all_polys.empty:
        all_polys.plot(ax=ax, color='#efefef', edgecolor='none', alpha=0.55, zorder=2)
    if not all_lines.empty:
        all_lines.plot(ax=ax, color='#cfcfcf', linewidth=0.35, alpha=0.8, zorder=2)
    if not all_points.empty:
        all_points.plot(ax=ax, color='#d9d9d9', markersize=8, alpha=0.8, zorder=2)

    # Highlight increases in strong red
    gt_inc = inc.geometry.geom_type.astype(str)
    inc_polys = inc[gt_inc.str.contains('Polygon', na=False)]
    inc_lines = inc[gt_inc.str.contains('LineString', na=False)]
    inc_points = inc[gt_inc.str.contains('Point', na=False)]

    if not inc_polys.empty:
        inc_polys.plot(ax=ax, color='#ff2d2d', edgecolor='#990000', linewidth=0.20, alpha=0.90, zorder=3)
    if not inc_lines.empty:
        inc_lines.plot(ax=ax, color='#ff0000', linewidth=1.6, alpha=0.95, zorder=4)
    if not inc_points.empty:
        inc_points.plot(ax=ax, color='#ff0000', markersize=26, alpha=0.95, zorder=5)

    ax.set_title(title, fontsize=13)
    ax.text(
        0.01, 0.01,
        f"Highlighted assets: {int(inc_mask.sum()):,} | Total increase: {format_usd_readable(total_increase)}",
        transform=ax.transAxes,
        ha='left', va='bottom',
        fontsize=10,
        bbox=dict(facecolor='white', edgecolor='#bbbbbb', alpha=0.9)
    )
    ax.set_axis_off()
    plt.tight_layout()

    fig.savefig(output_png, dpi=300, bbox_inches='tight')
    print('Saved:', output_png)
    plt.show()
    return True


forest_increase_map_protection_highlight = map_out_dir / 'forest_associated_damage_increase_protection_highlight_red_all_sectors.png'
forest_increase_map_restoration_highlight = map_out_dir / 'forest_associated_damage_increase_restoration_highlight_red_all_sectors.png'

has_protection_highlight = plot_binary_increase_map(
    landslide_ead_map_layers,
    value_col='Forest_Associated_Increase_Protection_USD',
    title='Any Increase with Forests Highlighted in Red (Baseline > Deforestation)',
    output_png=forest_increase_map_protection_highlight,
)

has_restoration_highlight = plot_binary_increase_map(
    landslide_ead_map_layers,
    value_col='Forest_Associated_Increase_Restoration_USD',
    title='Any Restoration-Associated Increase Highlighted in Red (Reforestation > Baseline)',
    output_png=forest_increase_map_restoration_highlight,
)

if not has_restoration_highlight:
    no_map_note2 = output_damage_estimates / 'forest_associated_damage_increase_restoration_highlight_none.txt'
    no_map_note2.write_text('No restoration-associated increase locations to highlight in red.\n', encoding='utf-8')
    print('Saved:', no_map_note2)
```

```python
# Intersect forest-associated increase areas with land-use map
landuse_path = base_path / 'dphil_common_cross_cutting/common_incoming_data/landcover/2013_landcover/2013_landuse_landcover.gpkg'
landuse_layer = '2013_landuse_landcover'

if not landuse_path.exists():
    raise FileNotFoundError(f'Missing land-use dataset: {landuse_path}')

landuse = gpd.read_file(landuse_path, layer=landuse_layer)
if landuse.crs is not None and str(landuse.crs).upper() != 'EPSG:3448':
    landuse = landuse.to_crs('EPSG:3448')

if 'Classify' not in landuse.columns:
    raise KeyError("Land-use layer missing 'Classify' column")
if 'LU_CODE' not in landuse.columns:
    landuse['LU_CODE'] = pd.NA

landuse = landuse[['Classify', 'LU_CODE', 'geometry']].copy()
landuse = landuse[landuse.geometry.notnull()].copy()
landuse['geometry'] = landuse.geometry.buffer(0)
landuse = landuse[~landuse.geometry.is_empty].copy()


def allocate_increase_to_landuse(increase_col, prefix):
    source = landslide_ead_map_layers.loc[
        landslide_ead_map_layers[increase_col] > 0,
        ['Sector', 'Subsector', 'Asset', 'Layer', 'Asset_ID', increase_col, 'geometry']
    ].copy()

    # Save empty outputs if none.
    out_csv = output_damage_estimates / f'landslide_{prefix}_landuse_class_summary_usd.csv'
    out_gpkg = output_damage_estimates / f'landslide_{prefix}_landuse_intersection.gpkg'

    if source.empty:
        empty = pd.DataFrame(columns=[
            'Classify', 'LU_CODE',
            'Allocated_Increase_USD',
            'Affected_Asset_Count',
            'Intersection_Length_km',
            'Intersection_Area_sqkm',
            'Affected_Point_Count'
        ])
        empty.to_csv(out_csv, index=False)
        print(f'No positive {increase_col}; wrote empty summary:', out_csv)
        return empty, None

    source = source[source.geometry.notnull()].copy()
    source['geometry'] = source.geometry.buffer(0)
    source = source[~source.geometry.is_empty].copy()

    geom_type = source.geometry.geom_type.astype(str)
    points = source[geom_type.str.contains('Point', na=False)].copy()
    lines = source[geom_type.str.contains('LineString', na=False)].copy()
    polys = source[geom_type.str.contains('Polygon', na=False)].copy()

    pieces = []

    # Points: spatial join with land-use polygons.
    if not points.empty:
        pjoin = gpd.sjoin(
            points,
            landuse[['Classify', 'LU_CODE', 'geometry']],
            how='left',
            predicate='within'
        )
        pjoin['Allocated_Increase_USD'] = pjoin[increase_col]
        pjoin['Intersection_Length_km'] = 0.0
        pjoin['Intersection_Area_sqkm'] = 0.0
        pjoin['Affected_Point_Count'] = 1
        pjoin['Asset_Geom_Type'] = 'Point'
        pieces.append(pjoin[['Sector', 'Subsector', 'Asset', 'Layer', 'Asset_ID', 'Classify', 'LU_CODE',
                             'Allocated_Increase_USD', 'Intersection_Length_km', 'Intersection_Area_sqkm',
                             'Affected_Point_Count', 'Asset_Geom_Type', 'geometry']])

    # Lines: intersect with land-use polygons and allocate by length share.
    if not lines.empty:
        lines = lines.copy()
        lines['orig_measure'] = lines.geometry.length.replace(0, pd.NA)
        line_ov = gpd.overlay(
            lines,
            landuse[['Classify', 'LU_CODE', 'geometry']],
            how='intersection',
            keep_geom_type=False,
        )
        if not line_ov.empty:
            line_ov = line_ov[line_ov.geometry.notnull()].copy()
            line_ov = line_ov[~line_ov.geometry.is_empty].copy()
            line_ov['piece_measure'] = line_ov.geometry.length
            line_ov['weight'] = (line_ov['piece_measure'] / line_ov['orig_measure']).fillna(0.0)
            line_ov['Allocated_Increase_USD'] = line_ov[increase_col] * line_ov['weight']
            line_ov['Intersection_Length_km'] = line_ov['piece_measure'] / 1000.0
            line_ov['Intersection_Area_sqkm'] = 0.0
            line_ov['Affected_Point_Count'] = 0
            line_ov['Asset_Geom_Type'] = 'LineString'
            pieces.append(line_ov[['Sector', 'Subsector', 'Asset', 'Layer', 'Asset_ID', 'Classify', 'LU_CODE',
                                   'Allocated_Increase_USD', 'Intersection_Length_km', 'Intersection_Area_sqkm',
                                   'Affected_Point_Count', 'Asset_Geom_Type', 'geometry']])

    # Polygons: intersect with land-use polygons and allocate by area share.
    if not polys.empty:
        polys = polys.copy()
        polys['orig_measure'] = polys.geometry.area.replace(0, pd.NA)
        poly_ov = gpd.overlay(
            polys,
            landuse[['Classify', 'LU_CODE', 'geometry']],
            how='intersection',
            keep_geom_type=False,
        )
        if not poly_ov.empty:
            poly_ov = poly_ov[poly_ov.geometry.notnull()].copy()
            poly_ov = poly_ov[~poly_ov.geometry.is_empty].copy()
            poly_ov['piece_measure'] = poly_ov.geometry.area
            poly_ov['weight'] = (poly_ov['piece_measure'] / poly_ov['orig_measure']).fillna(0.0)
            poly_ov['Allocated_Increase_USD'] = poly_ov[increase_col] * poly_ov['weight']
            poly_ov['Intersection_Area_sqkm'] = poly_ov['piece_measure'] / 1_000_000.0
            poly_ov['Intersection_Length_km'] = 0.0
            poly_ov['Affected_Point_Count'] = 0
            poly_ov['Asset_Geom_Type'] = 'Polygon'
            pieces.append(poly_ov[['Sector', 'Subsector', 'Asset', 'Layer', 'Asset_ID', 'Classify', 'LU_CODE',
                                   'Allocated_Increase_USD', 'Intersection_Length_km', 'Intersection_Area_sqkm',
                                   'Affected_Point_Count', 'Asset_Geom_Type', 'geometry']])

    if not pieces:
        empty = pd.DataFrame(columns=[
            'Classify', 'LU_CODE',
            'Allocated_Increase_USD',
            'Affected_Asset_Count',
            'Intersection_Length_km',
            'Intersection_Area_sqkm',
            'Affected_Point_Count'
        ])
        empty.to_csv(out_csv, index=False)
        print(f'No intersections created for {increase_col}; wrote empty summary:', out_csv)
        return empty, None

    inter = gpd.GeoDataFrame(pd.concat(pieces, ignore_index=True), geometry='geometry', crs='EPSG:3448')

    # Save map-ready intersection layer.
    inter.to_file(out_gpkg, driver='GPKG')
    print('Saved intersection layer:', out_gpkg)

    # Summarise by land-use class.
    summary = (
        inter.groupby(['Classify', 'LU_CODE'], dropna=False, as_index=False)
        .agg(
            Allocated_Increase_USD=('Allocated_Increase_USD', 'sum'),
            Affected_Asset_Count=('Asset_ID', lambda x: x.astype(str).nunique()),
            Intersection_Length_km=('Intersection_Length_km', 'sum'),
            Intersection_Area_sqkm=('Intersection_Area_sqkm', 'sum'),
            Affected_Point_Count=('Affected_Point_Count', 'sum'),
        )
        .sort_values('Allocated_Increase_USD', ascending=False)
        .reset_index(drop=True)
    )

    summary['Allocated_Increase_Readable'] = summary['Allocated_Increase_USD'].apply(format_usd_readable)
    summary.to_csv(out_csv, index=False)
    print('Saved summary:', out_csv)

    return summary, inter


protection_landuse_summary, _ = allocate_increase_to_landuse(
    'Forest_Associated_Increase_Protection_USD',
    'forest_associated_increase_protection'
)

restoration_landuse_summary, _ = allocate_increase_to_landuse(
    'Forest_Associated_Increase_Restoration_USD',
    'forest_associated_increase_restoration'
)

print('Top land-use classes for protection-associated increases:')
display(protection_landuse_summary.head(20))

print('Top land-use classes for restoration-associated increases:')
display(restoration_landuse_summary.head(20))
```
