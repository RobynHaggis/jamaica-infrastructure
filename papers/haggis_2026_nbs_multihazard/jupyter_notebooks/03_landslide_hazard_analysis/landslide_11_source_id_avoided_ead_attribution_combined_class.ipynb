# Landslide combined-class step 11: source-ID avoided EAD attribution

This notebook attributes landslide avoided EADs from forest protection and reafforestation to source-zone IDs using the `cell_source_*_dominant.tif` runout-attribution rasters.

Method summary:

1. Reuse the existing combined-class split intersections from step 01.
2. Sample `cell_source_{Scenario}_rp{ReturnPeriod}_dominant.tif` at each split asset cell.
3. Recompute return-period direct damages using the same minimum/maximum assumptions as step 02.
4. Aggregate direct damages by `Source_ID`, scenario, and return period.
5. Calculate source-level EAD and avoided EAD for forest protection and reafforestation.

`Source_ID = 0` is retained in tabular quality-control outputs as `Attribution_Status = 'unattributed_source_id_0'`. Geospatial outputs include only positive source IDs joined to `source_zones_Baseline.gpkg`.
